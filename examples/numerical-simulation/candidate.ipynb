# 2-D heat equation: low-rank reconstruction

Solves on a 128x128 grid, then reconstructs the field from its leading singular vectors.

```python
N = 128
DT = 1e-4
RANK = 10
u0 = initial_condition(N)
```

```python
u, info = solve_heat(u0, dt=DT, steps=4000)
print(f'Converged after {info.iterations} iterations')
print(f'Max residual = {info.residual:.1e}')
```

```python
u_r = reconstruct(u, rank=RANK)
err = np.linalg.norm(u - u_r) / np.linalg.norm(u)
print(f'Relative reconstruction error = {err:.4f}')
```
